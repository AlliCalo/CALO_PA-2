```python
import numpy as np
```

# Problem A


```python
np.random.seed(2112) #initialize 5x5 array
X = np.random.randint(10, 101, size=(5, 5))
```

```python
mean_X = np.mean(X) #normalizing the given array
std_X = np.std(X)
X_normalized = (X - mean_X) / std_X
```

```python
print("Original Array X:")
print(X)
print("\nNormalized Array X_normalized:")
print(X_normalized)
print("Mean of X_normalized (should be 0):", X_normalized.mean())
print("Standard Deviation of X_normalized (should be 1):", X_normalized.std())
```

```python
np.save("X_normalized.npy", X_normalized)
```

# Problem B

```python
C = (np.arange(1, 101) ** 3).reshape(10, 10) #first 100 positive integers is cubed and arranged to a 10x10
```

```python
divide_by_4 = C[C % 4 == 0] #values divisible by 4 is presented
```

```python
#printing of everything needed
print("Shape of C:", C.shape)
print("\nArray divide_by_4:\n", divide_by_4)
print("\nNumber of selected elements:", divide_by_4.size)
print("First element:", divide_by_4[0])
print("Last element:", divide_by_4[-1])
```

```python
np.save("div_by_4.npy", divide_by_4) #saves the result
```

# Problem C

```python
S = (np.arange(1, 37) ** 2).reshape(6, 6) #first 36 positive integers is arranged in a 6x6 array
```

```python
S_mean = np.mean(S) #computes the mean of all elements
```

```python
above_mean = S[S > S_mean] #filters elements greater than mean of S
```

```python
#Display required
print("Array S:\n", S)
print("\nS_mean:", S_mean)
print("\nArray above_mean:\n", above_mean)
print("\nNumber of selected elements:", above_mean.size)
print("First element:", above_mean[0])
print("Last element:", above_mean[-1])
```

```python
np.save("above_mean.npy", above_mean) #saves the result
```
